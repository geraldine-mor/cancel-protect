```python
import pandas as pd
df = pd.read_csv("outputs/datasets/cleaned/HotelBookingsValid.csv")
df.head()
```

```python
grouped = df.groupby(df["hotel"])
grouped.head()
```

```python
city_df = df[df["hotel"] == "City Hotel"].copy()
city_df.head()
```

```python
resort_df = df[df["hotel"] == "Resort Hotel"].copy()
resort_df.head()
```

```python
resort_df["reserved_room_type"].value_counts().sort_index()
```

```python
city_df["reserved_room_type"].value_counts().sort_index()
```

```python
city_family = city_df[(city_df["children"] > 0) | (city_df["babies"] > 0)]
city_family["reserved_room_type"].value_counts()
```

```python
city_df["family_booking"] = ((city_df["children"] > 0) | (city_df["babies"] > 0))
city_df["total_guests"] = city_df["adults"] + city_df["children"] + city_df["babies"]

room_summary = (
    city_df.groupby("reserved_room_type")
      .agg(
          bookings=("reserved_room_type", "size"),
          avg_adults=("adults", "mean"),
          avg_children=("children", "mean"),
          avg_babies=("babies", "mean"),
          avg_guests=("total_guests", "mean"),
          pct_family=("family_booking", "mean"),
          avg_adr=("adr", "mean")
      )
)

room_summary["pct_family"] *= 100

room_summary = room_summary.round({
    "avg_adults":2,
    "avg_children":2,
    "avg_babies":2,
    "avg_guests":2,
    "pct_family":1,
    "avg_adr":2
}).sort_values("bookings", ascending=False).reset_index()
```

```python
room_summary
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, (ax1, ax2, ax3) = plt.subplots(
    ncols=1,
    nrows=3,
    figsize=(15,8)
)

sns.barplot(city_df, x=city_df["reserved_room_type"], y=city_df["adults"], ax=ax1)
sns.barplot(city_df, x=city_df["reserved_room_type"], y=city_df["children"], ax=ax2)
sns.barplot(city_df, x=city_df["reserved_room_type"], y=city_df["babies"], ax=ax3)

plt.tight_layout()
plt.show()
```
